```python
# 1. Importamos la librería pandas y le asignamos el apodo 'pd'
# ¿Por qué?: 'pd' es el estándar en la industria. Nos ahorra escribir 'pandas' cada vez que usemos sus herramientas.
import pandas as pd

# 2. Leemos el archivo CSV y lo guardamos en una variable llamada 'df'
# ¿Por qué?: Usamos la ruta relativa '../data/raw/train.csv'. 
# Los dos puntos '..' le dicen a Python "sal de la carpeta actual (notebooks)", y luego entra a data/raw.
df = pd.read_csv('../data/raw/train.csv')

# 3. Mostramos las primeras 5 filas de nuestra tabla (DataFrame)
# ¿Por qué?: Para verificar rápidamente que el archivo cargó bien, sin congelar la pantalla imprimiendo 9,800 filas.
df.head()
```

```python
# 1. Revisamos la información general y la estructura de nuestra tabla
# ¿Por qué?: Necesitamos saber exactamente cuántas filas tenemos, de qué tipo es cada columna 
# y, lo más importante, detectar si hay "huecos" (datos nulos o vacíos) que debamos limpiar.
df.info()
```

```python
# 1. Convertimos las columnas de texto a formato de fecha especial de Python
# ¿Por qué?: Para poder hacer cálculos con tiempos (ej. restar Ship Date menos Order Date)
# y agrupar las ventas por mes o año más adelante en nuestro análisis.
# Usamos format='%d/%m/%Y' porque en tus requisitos notaste que vienen como Día/Mes/Año.
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d/%m/%Y')

# 2. Mostramos de nuevo la estructura para verificar el cambio
# ¿Por qué?: Queremos ver la lista actualizada y confirmar que el problema se solucionó.
df.info()
```

```python
# 1. Rellenamos los espacios vacíos (nulos) con un cero (0)
# ¿Por qué?: Para mantener esas 11 filas de ventas intactas sin que la base de datos nos dé error luego.
df['Postal Code'] = df['Postal Code'].fillna(0)

# 2. Convertimos los códigos postales a números enteros y luego a texto (string)
# ¿Por qué?: Un código postal no es una cantidad que vayas a sumar o multiplicar. 
# Pasarlo a 'int' le quita el decimal (.0) y pasarlo a 'str' lo convierte en una etiqueta de texto.
df['Postal Code'] = df['Postal Code'].astype(int).astype(str)

# 3. Verificamos la cantidad de nulos que quedan en cada columna
# ¿Por qué?: El método isnull().sum() nos da un conteo rápido. Queremos ver puro cero.
df.isnull().sum()
```

```python
# 1. Verificamos si hay filas exactamente duplicadas (clones) en nuestra tabla
# ¿Por qué?: Los sistemas a veces fallan y registran una misma venta dos veces. 
# Si existen, tendríamos que borrarlas con df.drop_duplicates(), pero primero hay que contarlas.
duplicados = df.duplicated().sum()
print(f"Filas duplicadas encontradas: {duplicados}")

# 2. Convertimos los nombres de las columnas a minúsculas y cambiamos espacios por guiones bajos
# ¿Por qué?: Para aplicar el formato estándar (snake_case) que facilita la escritura de código SQL.
# Ejemplo: 'Order Date' se convertirá en 'order_date'.
df.columns = df.columns.str.lower().str.replace(' ', '_').str.replace('-', '_')

# 3. Imprimimos la lista de las nuevas columnas para verificar el cambio
# ¿Por qué?: Para asegurarnos visualmente de que ya no hay mayúsculas ni espacios.
print("\nNuevos nombres de columnas:")
print(df.columns.tolist())
```

```python
# 1. Guardamos nuestra tabla limpia en un nuevo archivo CSV
# ¿Por qué?: Protegimos la data original (raw). Ahora guardamos la versión procesada en la carpeta 'clean'.
# Usamos index=False para que Pandas no nos agregue una columna inútil con la numeración de las filas.
df.to_csv('../data/clean/train_clean.csv', index=False)

# 2. Imprimimos un mensaje de éxito
# ¿Por qué?: Siempre es buena práctica que el código nos confirme que terminó su trabajo sin errores.
print("¡Archivo limpio guardado con éxito en data/clean/train_clean.csv!")
```
