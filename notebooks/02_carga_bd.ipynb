```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Cargar el CSV limpio de la Etapa 2
df = pd.read_csv('../data/clean/train_clean.csv')

# 2. Separar la data para cada tabla (Normalización con Pandas)
# Usamos drop_duplicates() para extraer valores únicos y no insertar repetidos
df_customers = df[['customer_id', 'customer_name', 'segment']].drop_duplicates(subset=['customer_id'])
df_products = df[['product_id', 'category', 'sub_category', 'product_name']].drop_duplicates(subset=['product_id'])
df_orders = df[['order_id', 'order_date', 'ship_date', 'ship_mode', 'country', 'city', 'state', 'postal_code', 'region', 'customer_id']].drop_duplicates(subset=['order_id'])

# La tabla de detalles guarda cada línea individual de compra (no borramos duplicados aquí)
df_order_details = df[['row_id', 'order_id', 'product_id', 'sales']]

# 3. Crear el "puente" de conexión hacia PostgreSQL
# Sintaxis: postgresql://usuario:clave@localhost:puerto/nombre_bd
motor_bd = create_engine('postgresql+psycopg2://postgres:apdo30@localhost:5432/superstore_db')

# 4. Inyectar los datos en orden (primero los catálogos principales, luego las tablas con llaves foráneas)
print("Iniciando carga de Clientes...")
df_customers.to_sql('customers', con=motor_bd, if_exists='append', index=False)

print("Iniciando carga de Productos...")
df_products.to_sql('products', con=motor_bd, if_exists='append', index=False)

print("Iniciando carga de Pedidos (Orders)...")
df_orders.to_sql('orders', con=motor_bd, if_exists='append', index=False)

print("Iniciando carga de Detalle de Pedidos...")
df_order_details.to_sql('order_details', con=motor_bd, if_exists='append', index=False)

print("¡Base de datos cargada con éxito! Revisa DBeaver.")
```
